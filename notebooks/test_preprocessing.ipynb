```python
from src.config import INPUT_FILE, MATERIAL_FILE
from src.data_loader import load_raw_data


df_input, df_material = load_raw_data(
    INPUT_FILE,
    MATERIAL_FILE
)

print("Input shape:", df_input.shape)
print("Material shape:", df_material.shape)

print("\nInput columns:")
print(df_input.columns.tolist())

print("\nMaterial columns:")
print(df_material.columns.tolist())
```

```python
from src.preprocessing_common import clean_input

df_input_clean = clean_input(df_input)

print("Rows before:", len(df_input))
print("Rows after:", len(df_input_clean))

print(
    "Negative values after cleaning:",
    (df_input_clean["Masse"] < 0).sum()
)

print(
    "Zero values before cleaning:",
    (df_input["Masse"] == 0).sum()
)

print(
    "Zero values after cleaning:",
    (df_input_clean["Masse"] == 0).sum()
)

print(
    "Missing values after cleaning (block zero + negative):",
    df_input_clean["Masse"].isna().sum()
)
```

```python
from src.config import MATERIALS
from src.config import INPUT_FILE, MATERIAL_FILE
from src.data_loader import load_raw_data
from src.preprocessing_common import clean_materials

df_input, df_material = load_raw_data(INPUT_FILE, MATERIAL_FILE)

df_material_clean = clean_materials(
    df_material,
    MATERIALS
)

print("Before Clean Shape:", df_material.shape)
print("Clean Shape:", df_material_clean.shape)

print("\nColumns:")
print(df_material_clean.columns.tolist())

print("\nDate dtype:")
print(df_material_clean["Date"].dtype)

print("\nDate range:")
print(
    df_material_clean["Date"].min(),
    "to",
    df_material_clean["Date"].max()
)

print("\nDuplicate dates:")
print(df_material_clean["Date"].duplicated().sum())
```

```python
from src.pipeline_data import run_data_preparation_pipeline
from src.config import DATA_MODE
df_base = run_data_preparation_pipeline(DATA_MODE)

print("Shape:", df_base.shape)
print("Date range:", df_base["Date"].min(), "to", df_base["Date"].max())
print("Duplicate dates:", df_base["Date"].duplicated().sum())
print("Weekend rows:", (df_base["Date"].dt.dayofweek >= 5).sum())

print("Missing Input:", df_base["Masse"].isna().sum())
print("Rows with all Materials missing:", df_base[MATERIALS].isna().all(axis=1).sum())

print(df_base[[
    "Date",
    "is_holiday",
    "is_school_holiday"
]].head())

print(
    "Public holidays:",
    df_base["is_holiday"].sum()
)

print(
    "School holiday days:",
    df_base["is_school_holiday"].sum()
)

print(
    "Missing is_holiday:",
    df_base["is_holiday"].isna().sum()
)

print(
    "Missing is_school_holiday:",
    df_base["is_school_holiday"].isna().sum()
)

print("\nColumns:")
print(df_base.columns.tolist())
```
